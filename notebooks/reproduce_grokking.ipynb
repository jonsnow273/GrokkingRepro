```python
import json
import matplotlib.pyplot as plt

with open("../results/history.json", "r") as f:
    history = json.load(f)

steps = history["step"]
train_acc = [a * 100 for a in history["train_acc"]]
val_acc = [a * 100 for a in history["val_acc"]]

print(f"Total logged points: {len(steps)}")
print(f"Final train accuracy: {train_acc[-1]:.1f}%")
print(f"Final validation accuracy: {val_acc[-1]:.1f}%")
```

```python
fig, ax = plt.subplots(figsize=(9,5))
ax.plot(steps, train_acc, label="train_accuracy", color="#2E86AB", linewidth=2)
ax.plot(steps, train_acc, label="val_accuracy", color="#E63946", linewidth=2)
ax.set_xlabel("optimization steps")
ax.set_ylabel("accuracy (%)")
ax.set_title("Grokking: Train vs Validation Accuracy (p=97, addition, weight_decay=1.0)")
ax.legend(loc="center right")
ax.grid(alpha=0.3)
ax.set_ylim(-5, 105)
plt.tight_layout()
plt.show()
```
